Inpatient Rehabilitation Facility - National Data
National data on the quality of patient care measures shown on Inpatient Rehabilitation Facility Compare.

Columns

**1. CMS Certification Number (CCN)**

* identifies the facility associated with the data. Because this dataset contains national-level data rather than individual facilities, this field is listed as “Nation.”

**2. Measure code**

* A unique code that identifies the specific CMS quality measure being reported. It combines the CMS measure ID with information identifying the type of score.

**3. Score**

* The national score or value reported for the corresponding quality measure.

**4. Footnote**

* Indicates whether a footnote or additional explanation applies to the reported measure. CMS currently reports no footnotes for the national data.

**5. Start Date**

* The beginning date of the reporting period used to calculate the measure and its score.

**6. End date**

* The ending date of the reporting period used to calculate the measure and its score.

**7. Measure Data Range**

* Shows the full reporting period associated with the measure and score. If the reporting period is split into separate time periods, the ranges are separated by a semicolon.

Emergency Department Volume and Capacity 2021–2023

* A unique identification number assigned to each healthcare facility.

**2. Facility Name**

* The name of the hospital or healthcare facility.

**3. County Name**
* The county where the facility is located.

**4. System**
* The healthcare system or organization that the facility belongs to.

**5. Year**
* The year associated with the facility reported data.

**6. Licensed Bed Size**
* The facility number of licensed hospital beds,
grouped into a bed-size range.

**7. Hospital Ownership**
* The type of organization that owns or operates the hospital, such as government or nonprofit.

**8. Urban/Rural Designation**
* Identifies whether the facility is located in an urban or rural area.

**9. Teaching Designation**
* Identifies whether the hospital is classified as a teaching or non-teaching facility.

**10. Category**
* The category used to classify the emergency department data or patient visits.

**11. Total ED Number of Visits**
* The total number of emergency department visits reported by the facility.

**12. ED Stations**
* The number of treatment stations available in the facility emergency department.

**13. ED Diagnosis Count**
* The number of emergency department diagnoses recorded for the specified category.

**14. Latitude**
* The geographic latitude of the healthcare facility.

**15. Longitude**
* The geographic longitude of the healthcare facility.

**16. Primary Care Shortage Area**
* Indicates whether the facility is located in an area designated as having a shortage of primary care services.

**17. Mental Health Shortage Area**

**18. Visits Per Station**
* The number of emergency department visits relative to the number of available ED treatment stations.

Unemployment dataset

This table contains data on the percent of the population in the labor force who are unemployed (unemployment rate), for California, its regions, counties, county divisions, cities/towns, and census tracts. Data is from the Local Area Unemployment Statistics (LAUS), Bureau of Labor Statistics (BLS), and the U.S. Census Bureau, American Community Survey (ACS).

**1. ind_id — Indicator ID**
* A unique identification number assigned to the indicator.

**2. ind_definition — Indicator Definition**
* A plain-language definition describing the indicator being measured.

**3. reportyear — Report Year**
* The year or range of years for which the indicator was reported.

**4. race_eth_code — Race/Ethnicity Code**
* A numeric code used to identify a specific race or ethnicity group.

**5. race_eth_name — Race/Ethnicity Name**
*The name of the race or ethnicity group associated with the data.

**6. geotype — Geographic Type**
* Identifies the type of geographic area represented, such as state, region, county, place, county division, or census tract.

**7. geotypevalue — Geographic Type Value**
* A geographic identification code used to identify the specific area, such as a FIPS, region, place, or census tract code.

**8. geoname — Geographic Name**
* The name of the geographic area represented by the record.

**9. county_name — County Name**
* The name of the county in which the geographic area is located.

**10. county_fips — County FIPS Code**
* A five-digit code used to uniquely identify a county.

**11. region_name — Region Name**
* The name of the MPO-based geographic region associated with the data.

**12. region_code — Region Code**
* A numeric code used to identify the MPO-based geographic region.

**13. Unemployment — Unemployed Population**
* The estimated number of people who are unemployed.

**14. Labor_force — Labor Force**
* The estimated number of people who are part of the labor force.

**15. Unemployment_rate — Unemployment Rate**
* The percentage of the labor force that is unemployed, calculated as unemployment divided by the labor force  100.

**16. ll_95ci — Lower 95% Confidence Interval**
* The lower limit of the 95% confidence interval for the unemployment rate.

**17. ul_95ci — Upper 95% Confidence Interval**
* The upper limit of the 95% confidence interval for the unemployment rate.

**18. se — Standard Error**
* The standard error of the unemployment rate estimate.

**19. rse — Relative Standard Error**
* A measure of the precision of the unemployment rate estimate relative to the size of the estimate.

**20. ca_decile — California Decile**
* Groups California geographic areas into 10 ranked categories based on their unemployment values.

**21. ca_rr — California Rate Ratio**
* Compares the unemployment rate for the geographic area with the overall California unemployment rate

**22. version — Version**
* The date and time indicating the version of the dataset.
